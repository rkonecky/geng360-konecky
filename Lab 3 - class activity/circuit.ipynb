```python
import numpy as np 

f = np.array([1, 100, 720, 5000]) # Hz 
```

```python
# set all the constants

Vin = 1 # Volts
C = 47e-9  # Farads 
R = 4.7e3 # Ohms 
```

```python
# calculate the output voltage at each frequency 
Zc = 1/(2*np.pi*f*C) # Ohms 
Z = np.sqrt(R**2 + Zc**2) # Ohms
Vout = Vin * (Zc/Z) # Volts
for freq, vout in zip(f, Vout):
    print(f"At {freq} Hz, Vout = {vout:.3f} V")
```
