## Displacement

$$\delta_{max} = \frac{FL^3}{3EI}$$

$$I = \frac{bh^3}{12}$$

```python
import numpy as np # You only need to import this in the first cell

F = ((100/1000) * 9.81) # N, Applied Force (converting from grams to Newtons) 
L = 0.311 # m, Beam Length
E = 68.9e9  # Pa, Young's Modulus for 6061 Aluminum
rho = 2700  # kg/m^3, density
h = 0.0025  # m, beam height 
b = 0.0125 # m, beam width

I = b * h**3 / 12  # m^4, moment of inertia ; recall that in python ** is how you raise something to a power (not ^)

delta = F * L**3 / (3 * E * I)

# f-string: {delta:.3g} shows delta with 3 significant figures (general format)
# other options: {delta:.3f} fixed 3 decimals, {delta:.2e} scientific, {delta:.0f} nearest int

print(f'Displacement is {delta:.3g} m') 

delta_mm = delta * 1000  # convert from m to mm
print(f'Displacement is {delta_mm:.3g} mm') 
```

## Stress

$$\sigma = \frac{y_{max} M}{I} = \frac{y_{max} F  L}{I}$$

```python
h = 0.0025  # m, beam height 
b = 0.0125 # m, beam width
F = ((10/1000) * 9.81) # N, Applied Force (converting from grams to Newtons)
L = 0.311  # m, beam length 

I = b * h**3 / 12  # m^4, moment of inertia

print(f'Moment of inertia is {I:.3g} m^4')

ymax = h / 2  # m, distance from neutral axis to outer fiber 

print(f'ymax is {ymax:.3g} m')

sigma_max_Pa = F * L * ymax / I  # Pa, max bending stress 

print(f'Stress is {sigma_max_Pa:.3g} Pa') 

sigma_max_psi = sigma_max_Pa * 0.000145038  # psi, max bending stress converted from Pa
print(f'Stress is {sigma_max_psi:.3g} psi')
```

## Strain

$$\epsilon = \frac{\sigma}{E}$$

```python
E = 68.9e9 # Pa Young's Modulus for 6061 Aluminum 

epsilon_max = sigma_max_Pa / E  # max strain (unitless) 

print(f'Strain is {epsilon_max:.3g}')
```

## Natural Frequency

$$f_n = \frac{K_n}{2\pi}\sqrt{\frac{EI}{\rho A L^4}}$$

```python
# Reminder to do these calculations in SI units,
# or else you have to account for a lbm to lbf
# unit conversion. No one wants to do that :) 

h = 0.0025  # m, beam height 
b = 0.0125 # m, beam width
Kn = 3.52 # constant
rho = 2700  # kg/m^3, density
A = b * h  # m^2, cross-sectional area
L = 0.311  # m, beam length 
E = 68.9e9  # Pa, Young's Modulus for 6061 Aluminum 
I = b * h**3 / 12  # m^4, moment of inertia 

fn = Kn / (2 * np.pi) * np.sqrt((E * I) / (rho * A * L**4))  # in Hz, fundamental frequency

print(f'Natural Frequency is {fn:.3g} Hz') 
```
