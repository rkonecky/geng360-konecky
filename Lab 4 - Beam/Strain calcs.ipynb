```python
# Convert data to V 

Vcc = 5.00

strain103g['StrainVout103g'] = (strain103g["strain (raw)"] / 128 ) * (Vcc / 16777215) 
#plot this to check it looks right
plt.figure()  # Create a new figure
sns.lineplot(x="Times (us)", y= "StrainVout103g", data=strain103g)
plt.title("Strain Voltage (10.3g) vs Time")
plt.xlabel("Time (us)")
plt.ylabel("Strain Voltage (V)")
plt.show()

Vin = 5.00 # V 
GF = 2.11 # Gauge factor for the strain gauge, this is a property of the material and geometry of the gauge.

# Compute baseline as average of first 100 samples (dimensionless ratio)
baseline_samples = 100
Vunstrained_ratio = strain103g['StrainVout103g'].iloc[:baseline_samples].mean() / Vin

strain103g['Vr103g'] = ( strain103g['StrainVout103g'] / Vin ) - Vunstrained_ratio 

strain103g['Vr103g'].head()

strain103g['strainfinal103g'] = (( strain103g['Vr103g'] * (-4) ) / (GF * (1 + (strain103g['Vr103g'] * 2)) ) )

strain103g['strainfinal103g'].head() 

# convert strain units to microstrain (1 microstrain = 1e-6 strain)
strain103g['strainfinal103g'] = strain103g['strainfinal103g'] * 1e6

# Zero the strain at the start (subtract initial value to make t=0 = 0)
initial_strain = strain103g['strainfinal103g'].iloc[0]
strain103g['strainfinal103g'] = strain103g['strainfinal103g'] - initial_strain

# plot stress vs time for 103g
plt.figure()  # Create a new figure
sns.lineplot(x="Times (us)", y= "strainfinal103g", data=strain103g)
plt.title("Strain (10.3g) vs Time")
plt.xlabel("Time (us)")
plt.ylabel("Strain (µε)")

mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection

# Click anywhere on your data to add a cursor
# Right-click to remove a cursor

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# USER PARAMETERS
# -----------------------------
FILENAME = r"C:\Users\bexte\Geng 360 shtuff\geng360-konecky\10.3gstrain.csv" # Update this to match your data file path

ADC_MAX = 2**24 - 1
GAIN = 128
GF = 2.0
V_IN = 5.0

# -----------------------------
# LOAD DATA
# -----------------------------
df = pd.read_csv(FILENAME)

# Extract columns
time_us = df["Times (us)"].values
raw_adc = df["strain (raw)"].values

# Convert time to seconds
time_s = time_us / 1e6

# -----------------------------
# CONVERT RAW → Vout/Vin
# -----------------------------
v_ratio = raw_adc / (ADC_MAX * GAIN)

# -----------------------------
# BASELINE (UNSTRAINED)
# Assume first ~100 samples are unstrained
# -----------------------------
baseline = np.mean(v_ratio[:100])

# Vr calculation
Vr = v_ratio - baseline

# -----------------------------
# STRAIN CALCULATION (NONLINEAR)
# ε = (-4 Vr) / (GF (1 + 2 Vr))
# -----------------------------
strain = (-4 * Vr) / (GF * (1 + 2 * Vr))

# Convert to microstrain
microstrain = strain * 1e6

# -----------------------------
# PLOT
# -----------------------------
plt.figure()

plt.plot(time_s, microstrain)

plt.xlabel("Time (s)")
plt.ylabel("Microstrain (µε)")
plt.title("Strain vs Time (10.3g Load)")

plt.grid()

plt.show()
```

```python
# -----------------------------
# USER INPUTS (MEASURE THESE)
# -----------------------------
FILENAME = r"C:\Users\bexte\Geng 360 shtuff\geng360-konecky\20.3gstrain.csv" # Update this to match your data file path

Vcc = 5.0          # Measure this with a multimeter!
Vin = Vcc          # Usually same, but confirm
GF = 2.0           # From strain gauge datasheet

ADC_MAX = 16777215  # 2^24 - 1
GAIN = 128

# -----------------------------
# LOAD DATA
# -----------------------------
df = pd.read_csv(FILENAME)

time_us = df["Times (us)"].values
raw = df["strain (raw)"].values

# Convert time → seconds
time_s = time_us / 1e6

# -----------------------------
# STEP 1: RAW → Vout (Volts)
# From lab:
# Vout = (Raw / 16777215) * (Vcc / 128)
# -----------------------------
Vout = (raw / ADC_MAX) * (Vcc / GAIN)

# -----------------------------
# STEP 2: Compute Vout/Vin
# -----------------------------
Vratio = Vout / Vin

# -----------------------------
# STEP 3: Find baseline (unstrained)
# Use first ~100 samples
# -----------------------------
Vratio_unstrained = np.mean(Vratio[:100])

# -----------------------------
# STEP 4: Vr
# -----------------------------
Vr = Vratio - Vratio_unstrained

# -----------------------------
# STEP 5: Strain (NONLINEAR EQN FROM PDF)
# ε = (-4 Vr) / (GF (1 + 2 Vr))
# -----------------------------
strain = (-4 * Vr) / (GF * (1 + 2 * Vr))

# -----------------------------
# STEP 6: Convert to µStrain
# -----------------------------
microstrain = strain * 1e6

# -----------------------------
# PLOT
# -----------------------------
plt.figure()

plt.plot(time_s, microstrain)

plt.xlabel("Time (s)")
plt.ylabel("Strain (µε)")
plt.title("Strain vs Time (20.3 g Load)")

plt.grid()

plt.show()
```
