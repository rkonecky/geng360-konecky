```python
# Convert data to V 

Vcc = 5.00

strain103g['StrainVout103g'] = (strain103g["strain (raw)"] / 128 ) * (Vcc / 16777215) 

Vunstrained = 2.74 # V 
Vin = 5.00 # V 
GF = 2.11 # Gauge factor for the strain gauge, this is a property of the material and geometry of the gauge.

strain103g['Vr103g'] = ( strain103g['StrainVout103g'] / Vin ) - Vunstrained

strain103g['Vr103g'].head()

strain103g['strainfinal103g'] = ( strain103g['Vr103g'] * (-4) ) / (GF * (1 + (strain103g['Vr103g'] * 2)) ) 

strain103g['strainfinal103g'].head() 

# convert strain units to microstrain (1 microstrain = 1e-6 strain)
strain103g['strainfinal103g'] = strain103g['strainfinal103g'] * 1e6

# plot stress vs time for 103g
plt.figure()  # Create a new figure
sns.lineplot(x="Times (us)", y= "strainfinal103g", data=strain103g)
plt.title("Strain (10.3g) vs Time")
plt.xlabel("Time (ms)")
plt.ylabel("Strain (10.3g)")

mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection

# Click anywhere on your data to add a cursor
# Right-click to remove a cursor

plt.show()
```
