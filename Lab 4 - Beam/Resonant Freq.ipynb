```python
# As alwayw, we start by importing our libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import sounddevice as sd

from scipy.signal import welch # We'll need this for power spectral density estimation 

# Allow for interactive plotting in Jupyter notebooks, comment out to turn this off. 
# The % sign is a Jupyter magic command that enables the use of widgets in the notebook.
# This requires installing the ipympl package.
%matplotlib widget 

# For interactive data point selection in plots, this library is useful.
import mplcursors
```

```python
VCC = 5.16  # set to 3.3 if your board is 3.3V
resfreq1 = pd.read_csv("C:\\Users\\bexte\\Geng 360 shtuff\\geng360-konecky\\resfreq1test1.csv") # Make sure to update this file name to match your data! 
resfreq1.head() #This will display the first few rows of the DataFrame
```

```python
plt.figure()  # Create a new figure
sns.lineplot(x="Time (ms)", y="Sensor 0 (raw)", data=resfreq1)
sns.lineplot(x="Time (ms)", y="Sensor 1 (raw)", data=resfreq1)
sns.lineplot(x="Time (ms)", y="Sensor 2 (raw)", data=resfreq1)
plt.title("Sensor vs Time")
plt.xlabel("Time (ms)")
plt.ylabel("Sensor (raw)")
#add a legend to differentiate the lines 
plt.legend(title="Sensors", labels=["Sensor 0", "Sensor 1", "Sensor 2"])


mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection

# Click anywhere on your data to add a cursor
# Right-click to remove a cursor

plt.show()
```
