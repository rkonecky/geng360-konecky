```python
# Conversion of raw sensor data to audio signal 

# This data was sampled at 250 Hz, so we set the sampling frequency.
fs_resfreq5 = 250

VCC = 5.00 # V from computer

resfreq5['AccelVolts'] = ((resfreq5['Sensor 0 (raw)'] / 1023) * VCC) - 2.5

avgfirst200 = resfreq5['AccelVolts'][:200].mean() # Average of the first 200 samples to find the baseline voltage

resfreq5['gs'] = resfreq5['AccelVolts'] / 0.3 # 270-330 mV/g is the sensitivity of the ADXL335 accelerometer


plt.figure()  # Create a new figure
sns.lineplot(x="Time (ms)", y="gs", data=resfreq5, color='magenta')
plt.title("Sensor vs Time")
plt.xlabel("Time (ms)")
plt.ylabel("Sensor (g)")
#add a legend to differentiate the lines 

mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection

# Click anywhere on your data to add a cursor
# Right-click to remove a cursor

plt.show()
```

```python
# Definte input parameters for Welch's method
segment_size = 512 # Size of each segment for Welch's method, should be a power of 2 (e.g. 256, 512, 1024, 2048, 4096)
overlap_pct = 40 # Percentage of overlap between segments, typically 25-75%
overlap_size = int(segment_size * overlap_pct / 100) # Calculate the number of samples to overlap between segments

# Compute Power Spectral Density using Welch's method
f, pxx = welch(resfreq5['gs'].values, 
               fs=fs_resfreq5, window="hann", nperseg=segment_size, noverlap=overlap_size)
#f, pxx = welch(flute['Scaled'], fs=fs_resfreq5, window="hann", nperseg=segment_size, noverlap=overlap_size)

# f is the frequency vector, pxx is the power spectral density estimate
# For more details, see the documentation https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.welch.html

plt.figure()
sns.lineplot(x=f, y=pxx, marker = '*')
plt.xscale('log')  # Logarithmic scale for frequency axis
plt.yscale('log')  # Logarithmic scale for power spectral density
plt.xlabel('Frequency (Hz)')
plt.ylabel('PSD Amplitude (Arbitrary Units)')
plt.title('Res Freq 5 Power Spectral Density')
plt.grid(True)
mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection
plt.show()
```
